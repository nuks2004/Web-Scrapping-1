```python
import requests 
```

```python
for i in range(1,11):
    a = requests.get(f"https://quotes.toscrape.com/page/{i}/")
    with open(f"Quotes/page{i}.html","w",encoding="utf-8")as f:
        f.write(a.text)
        print(f"Downloaded Page {i} Successfully")
        
```

```python
from bs4 import BeautifulSoup
```

```python
items =[]
for i in range(1,11):
    with open(f"Quotes/page{i}.html", encoding="utf-8") as f:
        content = f.read()
        soup = BeautifulSoup(content,"html.parser")
        divs = soup.select("div.quote")
        for div in divs:
            Quote=div.select_one("span.text").text.strip()
            author=div.select_one("small.author").text.strip()
            tags=[tag.text.strip() for tag in div.select("a.tag")]
            items.append([Quote,author,tags])
```

```python
import pandas as pd 

df= pd.DataFrame(items , columns=["Quote","author","tags"])
df.to_csv("data_complete.csv", index=False)
```
